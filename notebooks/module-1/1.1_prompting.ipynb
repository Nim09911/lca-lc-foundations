```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage

agent = create_agent(model="ollama:llama3.2")

question = HumanMessage(content="What's the capital of the moon?")

response = agent.invoke(
    {"messages": [question]}
)

print(response['messages'][1].content)
```

## Structured output

```python
from langchain_ollama import ChatOllama

model = llm = ChatOllama(
    model="llama3.2",
    num_ctx=16348,
    temperature=0.7,
)
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from pydantic import BaseModel, Field

class CapitalInfo(BaseModel):
    name: str = Field(description="The name of the capital city.")
    location: str = Field(description="The geographic or planetary location.")
    vibe: str = Field(description="The overall ambience, atmosphere, culture, or feel of the city. eg. futuristic, dystopian, utopian, cyberpubnk, etc.")
    economy: str = Field(description="The primary economic drivers or financial status. eg. mining, tourism, trade hub, etc.")



base_agent = create_agent(
    model=model,
    system_prompt="You are a science fiction writer, create a capital city at the users request.",
    response_format=CapitalInfo
)

agent = base_agent.with_config({ "recursion_limit": 32})

question = HumanMessage(content="What is the capital of The Moon?")

response = agent.invoke(
    {"messages": [question]}
)

response["structured_response"]
```

```python
response["structured_response"].name
```

```python
capital_info = response["structured_response"]

capital_name = capital_info.name
capital_location = capital_info.location

print(f"{capital_name} is a city located at {capital_location}")
```
